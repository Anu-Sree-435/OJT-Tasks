```python
import pandas as pd

df = pd.read_csv("flipkart_cleaned_data.csv")

df["overall_rating"] = pd.to_numeric(df["overall_rating"], errors="coerce")
df["discounted_price"] = pd.to_numeric(df["discounted_price"], errors="coerce")

df = df.dropna(subset=["overall_rating", "discounted_price"])

# coorelation between rating and price
correlation = df["discounted_price"].corr(df["overall_rating"])
print("Correlation between Price & Rating:", correlation)

# High-priced low-rated products
high_price_low_rating = df[
    (df["discounted_price"] > df["discounted_price"].mean()) &
    (df["overall_rating"] < 3)
]

# Low-priced high-rated (Value products)
low_price_high_rating = df[
    (df["discounted_price"] < df["discounted_price"].mean()) &
    (df["overall_rating"] >= 4)
]

high_price_low_rating.to_csv("high_price_low_rating.csv", index=False)
low_price_high_rating.to_csv("value_products.csv", index=False)

print("Task 3 Completed Successfully")
```
