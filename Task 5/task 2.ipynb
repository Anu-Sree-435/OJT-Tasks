```python
import pandas as pd

df = pd.read_csv("flipkart_cleaned_data.csv")

# Average price by category
avg_price = df.groupby("category")["discounted_price"].mean().sort_values(ascending=False)
print("\n Average Price by Category:\n", avg_price)

# Price range distribution
print("\nPrice Statistics:\n", df["discounted_price"].describe())

# identify Overpriced & Underpriced products
mean_price = df["discounted_price"].mean()
std_price = df["discounted_price"].std()

df["price_type"] = "Normal"
df.loc[df["discounted_price"] > mean_price + std_price, "price_type"] = "Overpriced"
df.loc[df["discounted_price"] < mean_price - std_price, "price_type"] = "Underpriced"

# Save result
df.to_csv("flipkart_price_analysis.csv", index=False)

print("Task 2 Completed: Price analytics saved")
```
