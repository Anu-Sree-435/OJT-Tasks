```python
import pandas as pd

df = pd.read_csv("flipkart_com-ecommerce_sample.csv")

df = df[[
    "product_name",
    "product_category_tree",
    "retail_price",
    "discounted_price",
    "overall_rating"
]]

df = df.drop_duplicates()

df["retail_price"] = pd.to_numeric(df["retail_price"], errors="coerce")
df["discounted_price"] = pd.to_numeric(df["discounted_price"], errors="coerce")

df = df.dropna(subset=["discounted_price", "overall_rating"])

df["category"] = df["product_category_tree"].apply(
    lambda x: x.split(">>")[0].replace("[\"", "").strip()
)

df = df.drop(columns=["product_category_tree"])

df.to_csv("flipkart_cleaned_data.csv", index=False)

print("Task 1 Completed: Cleaned dataset saved")
```
