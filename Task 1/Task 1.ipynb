```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
df= pd.read_csv("netflix_content_2023.csv")
```

```python
df.isnull().sum()
```

```python
df= df.dropna(axis=1, how="all")
```

```python
df.isnull().sum()
```

```python
df.head(10)  #Display 10 rows
```

```python
df['Hours Viewed'] = df['Hours Viewed'].replace({',': ''}, regex=True)
print("\nHours viewed for each title:")
print(df[['Title', 'Hours Viewed']])
```

```python
type_view = df.groupby('Content Type')['Hours Viewed'].sum()
plt.bar(type_view.index, type_view.values)
plt.title("Total Viewership Hours by Content Type (2023)")
plt.xlabel("Content Type")
plt.ylabel("Total Hours Viewed")
plt.show()
```

```python
type_view = df.groupby('Language Indicator')['Hours Viewed'].sum()
plt.bar(type_view.index, type_view.values)
plt.title("Total Viewership Hours by Language (2023)")
plt.xlabel("Content Type")
plt.ylabel("Total Hours Viewed")
plt.show()
```

```python
df['Release Month'] = pd.to_datetime(df['Release Date']).dt.month
plt.scatter(df['Release Month'], df['Hours Viewed'])
plt.title("Viewership Hours by Release Month (2023)")
plt.xlabel("Release Month")
plt.ylabel("Hours Viewed")
plt.show()
```

```python
top5 = df.nlargest(5, 'Hours Viewed')
print("\nTop 5 Titles based on Viewership Hours:\n", top5)
```

```python
plt.scatter(df['Release Month'], df['Hours Viewed'], c=pd.Categorical(df['Content Type']).codes)
plt.title("Viewership Trends by Content Type and Release Month (2023)")
plt.xlabel("Release Month")
plt.ylabel("Hours Viewed")
plt.show()
```
